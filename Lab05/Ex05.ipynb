a) Use the hmmlearn library to define the hidden Markov model and draw its state-transition diagram.

```python
import numpy as np
from hmmlearn import hmm
from graphviz import Digraph

states = ["Difficult", "Medium", "Easy"]
grades = ["VG", "G", "S", "U"]

pi = np.array([1/3, 1/3, 1/3])

# (D,M,E)
A = np.array([[0.0, 0.5,  0.5 ],
              [0.5, 0.25, 0.25],
              [0.5, 0.25, 0.25]])

B = np.array([[0.10, 0.20, 0.40, 0.30],
              [0.15, 0.25, 0.50, 0.10],
              [0.20, 0.30, 0.40, 0.10]])

model = hmm.CategoricalHMM(n_components=3, n_features=4, init_params="")
model.startprob_    = pi
model.transmat_     = A
model.emissionprob_ = B

#graph
g = Digraph("tests", format="png")
g.attr(rankdir="LR")

for i, s in enumerate(states):
    g.node(s, shape="circle")

for i, si in enumerate(states):
    for j, sj in enumerate(states):
        if A[i, j] > 0:
            g.edge(si, sj, label=f"{A[i, j]:.2f}")

g.render("transition_diagram", view=True)
```

b) Suppose you have observed the following sequence of grades: VG, VG, S, G, G, S, G, G, U, G, G, S. Determine
the most probable sequence of test difficulties corresponding to these observations and its probability.

```python
observed_grades = {"VG": 0, "G": 1, "S": 2, "U": 3}

model = hmm.CategoricalHMM(n_components=3, n_features=4, init_params="")
model.startprob_    = pi
model.transmat_     = A
model.emissionprob_ = B

seq = "VG VG S G G S G G U G G S".split()
obs = np.array([observed_grades[g] for g in seq]).reshape(-1, 1)   # column vector

log_prob, path = model.decode(obs, algorithm="viterbi")

print([states[i] for i in path])
print("log P =", log_prob, "  P =", np.exp(log_prob))
```
