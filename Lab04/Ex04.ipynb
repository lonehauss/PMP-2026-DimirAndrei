a) Vizualizaţi în pyAgrum graful reţelei Markov, iar apoi determinaţi clicile modelului.

```python
import pyagrum as gum
import pyagrum.lib.notebook as gnb

mn = gum.MarkovRandomField("MN")

for i in range(1, 6):
    mn.add(gum.LabelizedVariable(f"A{i}", f"A{i}", 2))

mn.addFactor({"A1", "A2"})
mn.addFactor({"A1", "A3"})
mn.addFactor({"A3", "A4"})
mn.addFactor({"A2", "A4", "A5"})

gnb.showMRF(mn)
```
